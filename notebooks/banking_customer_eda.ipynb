# 🏦 Banking Customer & Risk Analytics — EDA

**Notebook:** Exploratory Data Analysis
> Synthetic data — for portfolio demonstration only.


```python
import os, sys
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import warnings
warnings.filterwarnings('ignore')

# Adjust path if running from notebooks/ subfolder
BASE = os.path.dirname(os.path.abspath('.'))
DATA_PATH = os.path.join(BASE, 'data', 'processed', 'banking_customers_clean.csv')
if not os.path.exists(DATA_PATH):
    DATA_PATH = os.path.join('..', 'data', 'processed', 'banking_customers_clean.csv')

df = pd.read_csv(DATA_PATH)
df['Last_Transaction_Date'] = pd.to_datetime(df['Last_Transaction_Date'], errors='coerce')
print(f'Dataset shape: {df.shape}')
df.head()
```

```python
# Basic info
print('=== Data Types ===')
print(df.dtypes)
print('\n=== Missing Values ===')
print(df.isnull().sum()[df.isnull().sum() > 0])
print('\n=== Statistical Summary ===')
df.describe().round(2)
```

## 1. Customer Demographics

```python
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# Age distribution
axes[0,0].hist(df['Age'], bins=30, color='#3b82d4', edgecolor='white')
axes[0,0].set_title('Age Distribution'); axes[0,0].set_xlabel('Age')

# Gender
g_cnt = df['Gender'].value_counts()
axes[0,1].pie(g_cnt, labels=g_cnt.index, autopct='%1.1f%%', colors=['#3b82d4','#e74c3c','#27ae60'])
axes[0,1].set_title('Gender Distribution')

# Occupation
occ_cnt = df['Occupation'].value_counts()
axes[1,0].barh(occ_cnt.index, occ_cnt.values, color='#7c5cd8')
axes[1,0].set_title('Customers by Occupation')

# Account type
acc_cnt = df['Account_Type'].value_counts()
axes[1,1].bar(acc_cnt.index, acc_cnt.values, color='#27ae60', edgecolor='white')
axes[1,1].set_title('Customers by Account Type')

plt.tight_layout()
plt.savefig('../outputs/charts/demographics.png', dpi=150, bbox_inches='tight')
plt.show()
```

## 2. Income & Balance Analysis

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].hist(df['Monthly_Income'], bins=50, color='#e67e22', edgecolor='white')
axes[0].set_title('Monthly Income Distribution')
axes[0].set_xlabel('Monthly Income (INR)')

axes[1].hist(df['Account_Balance'], bins=50, color='#27ae60', edgecolor='white')
axes[1].set_title('Account Balance Distribution')
axes[1].set_xlabel('Account Balance (INR)')

plt.tight_layout()
plt.savefig('../outputs/charts/income_balance.png', dpi=150, bbox_inches='tight')
plt.show()

print('Income Summary:')
print(df['Monthly_Income'].describe().apply(lambda x: f'₹{x:,.0f}'))
```

## 3. Credit Score Analysis

```python
def credit_cat(s):
    if s < 500: return 'Very Poor'
    if s < 600: return 'Poor'
    if s < 700: return 'Fair'
    if s < 750: return 'Good'
    return 'Excellent'

df['Credit_Category'] = df['Credit_Score'].apply(credit_cat)
cc_cnt = df['Credit_Category'].value_counts().reindex(['Very Poor','Poor','Fair','Good','Excellent'])

fig, axes = plt.subplots(1, 2, figsize=(14, 5))
axes[0].hist(df['Credit_Score'], bins=40, color='#3b82d4', edgecolor='white')
axes[0].axvline(df['Credit_Score'].mean(), color='red', linestyle='--', label=f"Mean: {df['Credit_Score'].mean():.0f}")
axes[0].legend()
axes[0].set_title('Credit Score Distribution')

colors = ['#e74c3c','#e67e22','#f39c12','#27ae60','#2ecc71']
axes[1].bar(cc_cnt.index, cc_cnt.values, color=colors, edgecolor='white')
axes[1].set_title('Customers by Credit Category')
plt.tight_layout()
plt.savefig('../outputs/charts/credit_score.png', dpi=150, bbox_inches='tight')
plt.show()
```

## 4. Risk Distribution

```python
risk_cnt = df['Risk_Label'].value_counts()
colors = {'High':'#e74c3c','Medium':'#f39c12','Low':'#27ae60'}
clrs = [colors[r] for r in risk_cnt.index]

fig, axes = plt.subplots(1, 2, figsize=(12, 5))
axes[0].pie(risk_cnt, labels=risk_cnt.index, autopct='%1.1f%%', colors=clrs, startangle=90)
axes[0].set_title('Risk Distribution')

risk_by_cs = df.groupby(['Credit_Category','Risk_Label']).size().unstack(fill_value=0)
risk_by_cs = risk_by_cs.reindex(['Very Poor','Poor','Fair','Good','Excellent'])
risk_by_cs.plot(kind='bar', ax=axes[1], color=['#27ae60','#f39c12','#e74c3c'], edgecolor='white')
axes[1].set_title('Risk by Credit Category')
axes[1].tick_params(axis='x', rotation=30)
plt.tight_layout()
plt.savefig('../outputs/charts/risk_distribution.png', dpi=150, bbox_inches='tight')
plt.show()
```

## 5. Correlation Analysis

```python
numeric_cols = ['Age','Monthly_Income','Credit_Score','Account_Balance',
                'Loan_Amount','Debt_to_Income_Ratio','Missed_Payments',
                'Transaction_Amount','Customer_Satisfaction','Complaints']

corr = df[numeric_cols].corr()
plt.figure(figsize=(12, 9))
sns.heatmap(corr, annot=True, fmt='.2f', cmap='RdBu_r', center=0,
            square=True, linewidths=0.5)
plt.title('Correlation Heatmap — Key Features')
plt.tight_layout()
plt.savefig('../outputs/charts/correlation_heatmap.png', dpi=150, bbox_inches='tight')
plt.show()
```

## 6. Loan & Risk Analysis

```python
loan_df = df[df['Loan_Amount'] > 0].copy()
print(f'Loan customers: {len(loan_df):,} ({len(loan_df)/len(df)*100:.1f}%)')
print(f'Default rate: {loan_df["Default_Status"].mean()*100:.2f}%')
print(f'Average loan: ₹{loan_df["Loan_Amount"].mean():,.0f}')
print(f'Average DTI: {loan_df["Debt_to_Income_Ratio"].mean():.3f}')

fig, axes = plt.subplots(1, 2, figsize=(14, 5))
axes[0].hist(loan_df['Debt_to_Income_Ratio'], bins=40, color='#e74c3c', edgecolor='white')
axes[0].axvline(0.4, color='orange', linestyle='--', label='DTI=0.4')
axes[0].axvline(0.5, color='red', linestyle='--', label='DTI=0.5')
axes[0].legend()
axes[0].set_title('Debt-to-Income Ratio Distribution')

mp_def = loan_df.groupby(pd.cut(loan_df['Missed_Payments'],bins=[-1,0,2,5,100],
                                  labels=['0','1-2','3-5','6+']),
                          observed=True)['Default_Status'].mean().mul(100)
axes[1].bar(mp_def.index, mp_def.values, color=['#27ae60','#f39c12','#e67e22','#e74c3c'], edgecolor='white')
axes[1].set_title('Default Rate by Missed Payments')
axes[1].set_ylabel('Default Rate (%)')
plt.tight_layout()
plt.savefig('../outputs/charts/loan_risk.png', dpi=150, bbox_inches='tight')
plt.show()
```
